```python
import pandas as pod
import matplotlib.pyplot as plp 
```

# Analiza vrhov julijskih alp
Za projektno nalogo pri predmetu Uvod v programirajne sem analizirala podatke o vrhovih Julijskih alp. Podatke sem pridobila s spletne strani Hribi.net.


```python
hribi = pd.read_csv("hribi.csv")
hribi.head()
```

```python
hribi.info()
hribi.isna().sum()
```

```python
hribi = hribi.dropna(
    subset = ["ime", "visina"]
)
hribi.describe()

najvisji = hribi.sort_values(
    "visina",
    ascending = False
    ).head(10)

najvisji

najvisji_graf = najvisji.sort_values("visina)

najvisji_graf.plot(
    x = "ime",
    y = "visina",
    kind = "barh",
    legend = False    
)

plt.title("Deset najvišjih vrhov Julijskih alp")
plt.xlabel("Nadmorska visina v metrih")
plt.ylabel("Ime vrha")
plt.tight_layout()
plt.show()
```

```python
hribi["visina"].plot(
    kind = "hist",
    bins = 20,
    edgecolor = "pink"
)

plt.title("Porazdelitev višin vrhov")
plt.xlabel("Nadmorska višina v metrih")
plt.ylabel("Število vrhov")
plt.show()

najvec_poti = hribi.sort_values(
    "stevilo_poti",
    ascending = False
).head(10)

najvec_poti

najvec_poti_graf = najvec_poti.sort_values("stevilo_poti")

najvec_poti_graf.plot(
    x = "ime",
    y = "visina",
    kind = "barh",
    legend = False  
)

plt.title("Vrhovi z največ potmi")
plt.xlabel("Število poti")
plt.ylabel("Ime vrha")
plt.tight_layout()
plt.show()
```

```python
hribi.plot(
    x = "visina",
    y = "stevilo_poti",
    kind = "scatter"
)

plt.title("Povezava med visino vrha in številom poti")
plt.xlabel("Nadmorska višina v metrih")
plt.ylabel("Število poti")
plt.show()
```
